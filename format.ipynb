```python
import bibtexparser
from pathlib import Path

def format_reference(entry):
    """
    Format a bibliography entry into AAAI style reference 
    """
    authors = entry.get('author', '').split(' and ')
    formatted_authors = []
    for author in authors:
        names = author.split(', ')
        if len(names) == 2:
            formatted_authors.append(f"{names[0]}, {names[1]}")
        else:
            formatted_authors.append(author)
    
    if len(formatted_authors) == 1:
        author_str = formatted_authors[0]
    elif len(formatted_authors) == 2:
        author_str = f"{formatted_authors[0]} and {formatted_authors[1]}"
    else:
        author_str = ' and '.join(formatted_authors[:-1]) + f" and {formatted_authors[-1]}"

    title = entry.get('title', '').strip('{}')
    year = entry.get('year', '')
    
    if 'journal' in entry:
        journal = entry.get('journal', '')
        volume = entry.get('volume', '')
        number = entry.get('number', '')
        pages = entry.get('pages', '').replace('--', '-')
        
        ref = f"{author_str}. {year}. {title}. {journal}"
        if volume:
            ref += f" {volume}"
        if number:
            ref += f"({number})"
        if pages:
            ref += f":{pages}"
        ref += "."
        
    elif 'booktitle' in entry:
        booktitle = entry.get('booktitle', '')
        pages = entry.get('pages', '').replace('--', '-')
        
        ref = f"{author_str}. {year}. {title}. In {booktitle}"
        if pages:
            ref += f", {pages}"
        ref += "."
    
    else:
        ref = f"{author_str}. {year}. {title}."
    
    return ref

def process_bib_file(file_path):
    with open(file_path) as bibtex_file:
        bib_database = bibtexparser.load(bibtex_file)
    
    category = Path(file_path).stem
    papers = []
    
    for entry in bib_database.entries:
        paper = {
            'title': format_reference(entry),
            'tag': entry.get('tag', ''),
            'category': category
        }
        papers.append(paper)
    
    return papers
```

```python
papers = process_bib_file('computer vision.bib')
papers
```
